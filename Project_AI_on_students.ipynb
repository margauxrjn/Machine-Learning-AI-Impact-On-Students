Classifying into continuous and categorical features (continuous first because there is less, then all the others go into categorical)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

DATA_PATH = Path("ai_student_impact_dataset.csv")

df = pd.read_csv(DATA_PATH)

print("Dataset shape:", df.shape) #Numbers of rows and columns
print("\nColumn names:")
print(df.columns.tolist())
print("\nFirst five rows:")
display(df.head())

#Classifying : continuous features and categorical features
#Continuous first : there is less => categorical features are the rest
continuous_features = ["Pre_Semester_GPA","Weekly_GenAI_Hours","Traditional_Study_Hours","Post_Semester_GPA","Skill_Retention_Score"]
categorical_features = [col for col in df.columns if col not in continuous_features]

print("Continuous features:\n", continuous_features)
print("\n")
print("Categorical features:\n", categorical_features)
```

```python
from IPython.display import display
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score


# 1. Standardization
scaler = StandardScaler() # Z = (X-µ)/σ (the value minus the mean, on the standard deviation)
X_scaled = scaler.fit_transform(df[continuous_features]) #Application to continuous features

# 2. Optimizing the number of clusters (k)
K_range = range(2, 20)
inertias = [] # Within-cluster inertia 
silhouette_scores = [] # Should be HIGH -> Is the observation in a cluster with similar observations ? 
db_scores = [] # Davies-Bouldin scores, should be LOW (homogeneity and separability together)

for k in K_range:
    print(f"Computing k = {k}...")  # To follow live 
    #Initialization with K-means++ (farthest point from the first centroid chosen randomly)
    # With "K-means++", "auto" = 1 (number of times the algorithm runs with different centroids seeds)
    kmeans = KMeans(n_clusters=k, init='k-means++', n_init='auto') 
    labels = kmeans.fit_predict(X_scaled)

    inertias.append(kmeans.inertia_)
    # Limited sample_size=2000 so that Python do not got too slow
    silhouette_scores.append(silhouette_score(X_scaled, labels, sample_size=2000, random_state=None))
    db_scores.append(davies_bouldin_score(X_scaled, labels))

# 3. Graphs display (Elbow, Silhouette, Davies-Bouldin)
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Elbow Method (Within-cluster inertia)
axes[0].plot(K_range, inertias, marker='o', color='b')
axes[0].set_title("Elbow Method (Within-cluster inertia)")
axes[0].set_xlabel("Number of clusters (k)")
axes[0].set_ylabel("Within-cluster (Sum of squares")
axes[0].grid(True)

# Silhouette Score (better HIGH)
axes[1].plot(K_range, silhouette_scores, marker='s', color='g')
axes[1].set_title("Silhouette Score (Max = Best)")
axes[1].set_xlabel("Number of clusters (k)")
axes[1].set_ylabel("Silhouette Score")
axes[1].grid(True)

# Davies-Bouldin Index (better LOW)
axes[2].plot(K_range, db_scores, marker='^', color='r')
axes[2].set_title("Davies-Bouldin Index (Min = Best)")
axes[2].set_xlabel("Number of clusters (k)")
axes[2].set_ylabel("Davies-Bouldin Index")
axes[2].grid(True)

plt.tight_layout()
plt.savefig("kmeans_evaluation.png", dpi=300) # Saving the graph into the project
print("Graph saved  as 'kmeans_evaluation.png' !")
plt.show() #Display the graph

# Scores table
metrics_df = pd.DataFrame({
    "k": list(K_range),
    "Within-cluster inertia": inertias,
    "Silhouette_Score": silhouette_scores,
    "Davies_Bouldin_Index": db_scores
})
display(metrics_df)

# 4. Final model training 
# Optimal k given by the Elbow graph
optimal_k = 3

kmeans_final = KMeans(n_clusters=optimal_k, random_state=None, n_init='auto')
df["Cluster"] = kmeans_final.fit_predict(X_scaled)

print("\nStudent repartition per cluster :")
display(df["Cluster"].value_counts().to_frame("Number of students"))

print("\nCluster global profile :")
cluster_profile = df.groupby("Cluster")[continuous_features].mean().round(2)
display(cluster_profile)
```
